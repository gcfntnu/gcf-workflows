```python
def aggr_to_well_dataframes(adata, key='total_counts', well_layout=(8, 12), sample_map=None, layer=None, gene=None):
    """
    Converts an AnnData object into a well-based DataFrame representation by aggregating expression values.

    Parameters
    ----------
    adata : anndata.AnnData
        The annotated data matrix containing single-cell expression data.
    key : str, optional
        The column in `adata.obs` to use for populating the well layout (default is 'total_counts').
    well_layout : tuple of int, optional
        The shape of the well plate layout, given as (rows, columns). Default is (8, 12) for a 96-well plate.
    sample_map : dict, optional
        A mapping from well identifiers to sample IDs, applied only for the first barcode round.
    layer : str, optional
        The layer of `adata` from which to extract data. If None, defaults to `adata.X`.
    gene : str, optional
        If specified, extracts and logs the abundance of the given gene.

    Returns
    -------
    list of pandas.DataFrame
        A list of DataFrames, each representing a plate layout (one for each barcode round).
        Rows are labeled by well row (A, B, C, etc.), and columns by well number (1, 2, 3, ...).

    Raises
    ------
    ValueError
        If the specified `gene` is not found in `adata.var` under any recognized gene-symbol aliases.
        If well names do not follow the expected format (e.g., 'A1', 'B2').

    Notes
    -----
    - The function supports different well plate formats (e.g., 96-well, 384-well) via `well_layout`.
    - Gene symbols are extracted using an ordered list of aliases (`gene_symbol`, `gene_symbols`, etc.).
    - The function can compute QC metrics (`mt`, `ribo`, `hb`) if any of these exist in `adata.var`.
    """
    out = []
    
    for bc_round in [1, 2, 3]:
        # Aggregate counts by well
        bc = sc.get.aggregate(adata, by=f"bc{bc_round}_well", func=["sum"], layer=layer)

        # Map sample IDs if provided (only for bc_round 1)
        if sample_map and bc_round == 1:
            bc.obs['sample_id'] = [sample_map[i] for i in bc.obs.index if i in sample_map]

        # Calculate QC metrics for available qc_vars
        qc_vars = [var for var in ['mt', 'ribo', 'hb'] if var in adata.var.columns]
        if qc_vars:
            sc.pp.calculate_qc_metrics(bc, qc_vars=qc_vars, inplace=True, layer="sum")

        # Extract gene-specific information if provided
        if gene is not None:
            gene_symbols = next((a for a in ['gene_symbol', 'gene_symbols', 'gene_name', 'gene_names', 'name', 'names'] if a in adata.var.columns), None)
            if gene_symbols and gene in adata.var[gene_symbols].values:
                bc.obs[f'{gene}_abundance'] = np.log1p(sc.get.obs_df(bc, gene, gene_symbols=gene_symbols, layer="sum"))
                key = f'{gene}_abundance'
            else:
                raise ValueError(f"Gene '{gene}' not found in gene annotations.")

        # Define row and column labels based on well layout
        row_labels = [string.ascii_uppercase[i] for i in range(well_layout[0])]
        col_labels = [str(i+1) for i in range(well_layout[1])]
        well_df = pd.DataFrame(np.zeros(well_layout), index=row_labels, columns=col_labels)

        # Extract well positions
        ind = bc.obs.index.to_series().str.extract(r'(?P<row>[A-Z])(?P<col>\d+)', expand=True)
        if ind.isnull().any().any():
            raise ValueError("Well names do not follow expected format (e.g., 'A1', 'B2').")
        
        for n, row in ind.iterrows():
            well_df.at[row.row, row.col] = bc.obs[key][n]

        out.append(well_df)

    return out


def sns_wellmap(bc_rounds, sample_map=None, scale="m",
                title_key="total_counts", annot=True, save_path=None):
    """
    Generates a well plate heatmap visualization for three barcoding rounds.

    Parameters
    ----------
    bc_rounds : list of pandas.DataFrame
        A list of well plate DataFrames, one for each barcoding round.
    sample_map : dict, optional
        A mapping from well identifiers to sample names (not currently used).
    scale : {'m', 'k', None}, optional
        The scaling factor for values:
        - 'm' (default): scale by 1E6
        - 'k': scale by 1E4
        - None or other: no scaling
    title_key : str, optional
        Title label to describe the measured quantity (default is "total_counts").
    annot : bool, optional
        Whether to annotate heatmap cells with numerical values (default is True).
    save_path : str, optional
        If provided, saves the figure to the specified file path as a PNG.
        The saved image respects Matplotlib's rcParams for dpi and figure size.
    Returns
    -------
    tuple
        (fig, axes) where fig is the Matplotlib figure and axes is an array of subplots.

    Raises
    ------
    ValueError
        If a well plate has unexpected nonzero well counts.

    Notes
    -----
    - Uses a different colormap for each barcoding round (`Greens`, `Oranges`, `Blues`).
    - Assumes a 96-well or 12-well format; warns if more zero wells than expected.
    - Formats annotations based on the average count per well.
    """

    fig, axes = plt.subplots(
        nrows=3, sharex=True, 
        figsize=plt.rcParams.get("figure.figsize", (8, 8)), 
        layout="constrained",
        dpi=plt.rcParams.get("figure.dpi", 100)
    )
    cmaps = ["Greens", "Oranges", "Blues"]
    xtick_labels = False

    scale_factors = {"M": 1E6, "K": 1E4}
    scale = scale.upper()
    denom = scale_factors.get(scale, 1)
    scale_label = scale if scale in scale_factors else ""

    for i, well_df in enumerate(bc_rounds):
        n_above_zero = (well_df > 0).sum().sum()

        if n_above_zero not in [12, 96]:
            print(f"More zero wells than expected: n above zero: {n_above_zero}")

        avg_count = well_df.sum().sum() / max(n_above_zero, 1) / denom

        if i == 2:
            xtick_labels = "auto"

        ax = axes[i]
        cmap = cmaps[i]

        well_df_scaled = well_df / denom

        g = sns.heatmap(
            well_df_scaled,
            vmin=0,
            linecolor="#ECECEC",
            mask=well_df_scaled == 0,
            square=True,
            linewidths=5,
            annot=annot,
            fmt=".2f" if avg_count > 1 else "0.1f",
            robust=True,
            cbar=False,
            cmap=cmap,
            ax=ax,
        )

        g.set_yticklabels(g.get_yticklabels(), rotation=0,
                          horizontalalignment="right")

        kfmt = lambda x, pos: f"{x:1g}{scale_label}" if annot else None

        fig.colorbar(
            ax.collections[0], ax=ax, location="right", use_gridspec=False,
            format=kfmt, shrink=0.9, label=f"bc round {i+1}"
        )

        for _, spine in ax.spines.items():
            spine.set_visible(True)
            spine.set_color("gray")
            spine.set_linewidth(2)

    fig.subplots_adjust(hspace=0.15)
    fig.suptitle(f"{title_key} across the three barcoding steps\n", x=0.7)
    if save_path:
        fig.savefig(save_path, bbox_inches="tight")
    return fig, axes


```

```python
def sns_rankplot(adata, key='total_counts', key2=None, c='cell_probability', c2=None, subsample=False):
    obs = adata.obs.copy()

    # Ensure required columns exist
    required_cols = {key, key2, c, c2} - {None}
    missing_cols = required_cols - set(obs.columns)
    if missing_cols:
        raise ValueError(f"Missing columns in adata.obs: {missing_cols}")

    # Sort by primary key
    obs = obs.sort_values(by=[key], ascending=False)

    # Create rank index
    obs['x'] = np.arange(1, obs.shape[0] + 1)

    # Extract values
    x, y = obs['x'].values, obs[key].values
    col, col2 = obs[c].values, obs[c2].values if c2 else None
    y2 = obs[key2].values if key2 else None

    # Subsample for clarity
    if subsample:
        idx = np.arange(0, len(x), 10)
        x, y, col = x[idx], y[idx], col[idx]
        y2 = y2[idx] if y2 is not None else None
        col2 = col2[idx] if col2 is not None else None

    # Define cell categories
    call = np.where(col >= 0.5, "cell", "background")
    n_cells = (call=="cell").sum()
    n_umi_cell = y[n_cells]
    print(n_cells, n_umi_cell)

    # Define scatterplot markers based on c2
    if c2:
        unique_c2 = np.unique(col2)
        markers = {val: marker for val, marker in zip(unique_c2, ['o', 's', '^', 'D', 'X'])}
        marker_styles = [markers[val] for val in col2]

    # Create DataFrame for plotting
    df = pd.DataFrame({'Barcode Rank': x, 'UMI Count': y, 'Call': call})

    # Create jointplot
    g = sns.JointGrid(data=df, x="Barcode Rank", y="UMI Count",hue="Call", palette=["#2973B2", "#9ACBD0"])
    g.plot_joint(sns.scatterplot, s=15, alpha=0.5, linewidth=0)
    g.plot_marginals(sns.kdeplot, fill=True, alpha=0.5)
    # Set log scales
    g.ax_joint.set_xscale('log')
    g.ax_joint.set_yscale('log')

    # Move legend
    sns.move_legend(g.ax_joint, "upper left", bbox_to_anchor=(1, 1.2), title='')

    # **Hide only the marginal plots' axes**
    for gax in [g.ax_marg_x, g.ax_marg_y]:
        #gax.get_xaxis().set_visible(False)  
        gax.get_yaxis().set_visible(False) 
        
    #g.ax_joint.yaxis.set_tick_params(labelsize=10)
    #g.ax_joint.xaxis.set_tick_params(labelsize=10)

    if key2:
        g.ax_marg_y.remove()
        col2 = "#9DC08B"
        ax2 = g.ax_joint.twinx()
        g.ax_joint.set_zorder(2)
        ax2.set_zorder(1)
        if c2:
            for x_, y2_, m in zip(x, y2, marker_styles):
                ax2.semilogx(x_, y2_, color=col2, marker=m, alpha=0.6, linewidth=0, markersize=1)
        else:
            ax2.semilogx(x, y2, color=col2, alpha=0.6, marker='.', linewidth=0, markersize=1)
        ax2.set_ylabel(key2)
        ax2.grid(False)
        ax2.set_zorder(2)
        for _ax in [g.ax_joint, g.ax_marg_x]:
            _ax.set_zorder(3)
            _ax.patch.set_visible(False)
        

    return g
```
